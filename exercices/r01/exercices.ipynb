## 🟢 Exercice 1.1 — Broadcasting et surface d'énergie

**Contexte physique.** Le potentiel de Morse modélise la liaison chimique :
$$V(r) = D_e \left[1 - e^{-a(r-r_e)}\right]^2$$

**Tâche.** Pour une molécule diatomique hétéronucléaire (deux types d'atomes 
avec $D_e^{(1)} \neq D_e^{(2)}$), calculer la surface d'énergie potentielle 2D
$V(r_1, r_2)$ par broadcasting NumPy **sans boucle explicite**.

Paramètres : $D_e^{(1)} = 0.2$ Ha, $D_e^{(2)} = 0.15$ Ha, $a = 1.0$ $a_0^{-1}$,
$r_e = 1.5$ $a_0$. Grille : $r \in [0.8, 5.0]$ $a_0$, 300 points.


````python
# Cellule de solution (masquée par défaut — cliquer pour révéler)
import numpy as np

# Paramètres
De1, De2 = 0.2, 0.15
a, re    = 1.0, 1.5
r        = np.linspace(0.8, 5.0, 300)

# Potentiel de Morse pour chaque liaison
def V_morse(r, De, a, re):
    return De * (1 - np.exp(-a * (r - re)))**2

# Surface 2D par broadcasting
V1 = V_morse(r[:, None], De1, a, re)   # (300, 1)
V2 = V_morse(r[None, :], De2, a, re)   # (1, 300)
V_total = V1 + V2                       # (300, 300) — aucune boucle

print(f'Forme V_total : {V_total.shape}')
print(f'Minimum       : {V_total.min():.4f} Ha')
print(f'Position min  : r1={r[np.unravel_index(V_total.argmin(),V_total.shape)[0]]:.3f}, '
      f'r2={r[np.unravel_index(V_total.argmin(),V_total.shape)[1]]:.3f} a0')
````

## 🟡 Exercice 1.2 — Spectre et états propres du puits carré fini

**Contexte physique.** Le puits carré fini est le modèle minimal de la 
physique des semi-conducteurs (puits quantique). Pour un puits de profondeur 
$V_0$ et de largeur $2a$, le Hamiltonien discrétisé est une matrice tridiagonale.

**Tâche.**
1. Construire le Hamiltonien discrétisé sur une grille de $N = 500$ points.
2. Calculer les 5 premiers niveaux d'énergie par `np.linalg.eigh`.
3. Comparer avec les solutions transcendantes exactes (équation de dispersion).
4. Tracer les fonctions propres avec leur niveau d'énergie superposé au potentiel.

Paramètres (unités : $\hbar^2/2m = 1$) : $V_0 = 50$, $a = 1$, $N = 500$.


````python
# Solution Exercice 1.2
import numpy as np
import matplotlib.pyplot as plt

V0, a = 50.0, 1.0
N     = 500
L     = 4 * a     # domaine étendu
x     = np.linspace(-L, L, N)
dx    = x[1] - x[0]

# Potentiel
V = np.where(np.abs(x) <= a, 0.0, V0)

# Hamiltonien : -d²/dx² + V(x) (unités ℏ²/2m = 1)
diag    = 2/dx**2 + V
offdiag = -1/dx**2 * np.ones(N-1)
H = np.diag(diag) + np.diag(offdiag, 1) + np.diag(offdiag, -1)

# Diagonalisation
E, psi = np.linalg.eigh(H)

# États liés : E < V0
lies = E[E < V0]
print(f'États liés trouvés : {len(lies)}')
for k, e in enumerate(lies[:5]):
    print(f'  E_{k} = {e:.4f}')
````
